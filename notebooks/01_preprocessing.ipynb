# 2. Join mit manueller Tabelle

```python
path_excel = "../data/Excel/Metadaten_Lehrmaterialien.ods"
```

```python
df_excel = pd.read_excel(path_excel)
df_excel
```

```python
df_ = (
    data_preprocessing
    .left_join(
        df_projects_before,
        df_excel, # manual created table
        ["name_source", "id_name"]
    )
)
```

# 3. Vergabe einer eindeutigen ID
* Mehrfaches Nutzen vom pandas.index hat zu Problemen geführt, weshalb eigene ID erstellt wird

```python
df_["laufende_nr"] = range(len(df_))
```

```python
df_["document_id"] = df_["laufende_nr"].astype(str) + "_" + df_["title_name"] + "_" + df_["name_file"]
```

```python
df_
```

Cache result

```python
df_.to_parquet("../data/parquet/df_projects_before.parquet")
```

# 4. Filtern der Daten

* Jede Datei hat auf Basis seiner Endung eine Klassifizierung erhalten:
    * `primary_code`, `view_templates`, `configurations` und `unknown`
* Da nicht jede Datei bei der Filternung abgefangen werden konnte, wird nochmal manuell untersucht

```python
remove_extensions = config["remove_files"]["discard_extensions"]
```

```python
# Zeilen behalten, deren Wert NICHT in remove_dtypes enthalten ist
df_projects_after = df_[~df_["programming_language"].isin(remove_extensions)]
```

Vergleich Daten mit / ohne Datenvorverarbeitung

```python
print(f"Anzahl Vorher: {df_.shape[0]}")
print(f"Anzahl Nachher: {df_projects_after.shape[0]}")
print(f"Prozentuale Differenz: {round(1 - (df_projects_after.shape[0] / df_.shape[0]), 2) * 100}")
```
